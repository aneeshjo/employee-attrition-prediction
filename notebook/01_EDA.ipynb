Three features contain only a single unique value across all employee records:

- Over18

The following features will be removed during preprocessing:

- EmployeeCount

# Correlation Analysis

## Objective

Correlation analysis measures the strength and direction of the linear relationship between numerical variables.

Understanding these relationships helps us:

- Identify highly correlated features.
- Detect potential multicollinearity.
- Understand how employee-related characteristics interact.
- Guide feature selection and engineering.

It is important to note that correlation does not imply causation. The analysis is used as an exploratory tool to identify patterns that may be useful during model development.

```python
corr_matrix = df[numerical_columns].corr()

corr_matrix.head()
```

```python
import plotly.express as px

fig = px.imshow(
    corr_matrix,
    text_auto=".2f",
    aspect="auto",
    color_continuous_scale="RdBu_r",
    zmin=-1,
    zmax=1,
    title="Correlation Matrix of Numerical Features"
)

fig.update_layout(
    height=900,
    width=900
)

fig.show()
```

```python
corr_pairs = (
    corr_matrix
    .where(np.triu(np.ones(corr_matrix.shape), k=1).astype(bool))
    .stack()
    .reset_index()
)

corr_pairs.columns = [
    "Feature 1",
    "Feature 2",
    "Correlation"
]

strong_corr = (
    corr_pairs[
        abs(corr_pairs["Correlation"]) >= 0.70
    ]
    .sort_values(
        by="Correlation",
        ascending=False
    )
)

strong_corr
```

# Key Findings

No evidence of widespread multicollinearity is observed based on the initial correlation analysis.
